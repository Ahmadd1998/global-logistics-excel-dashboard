```python
#1. import library
import pandas as pd

#2.filepath yg digunakan
file_path = 'customer.xlsx'

#3.load semua file
df_customer = pd.read_excel(file_path, sheet_name='customer')
df_performance = pd.read_excel(file_path, sheet_name='logistics_performance')
df_shipment = pd.read_excel(file_path, sheet_name='shipment')
print("sukses")
```

```python
print('-- kolom di Customer --')
print(df_customer.columns.tolist())

print('\n-- kolom di Performance --')
print(df_performance.columns.tolist())

print('\n-- kolom di Shipment --')
print(df_shipment.columns.tolist())
```

```python
#Cleaning data duplikat
df_customer = df_customer.drop_duplicates()
df_performance = df_performance.drop_duplicates()
df_shipment = df_shipment.drop_duplicates()

#Cleaning baris "Unnamed" otomatis yang terbawa dari excel
df_customer = df_customer.loc[:, ~df_customer.columns.str.contains('^Unnamed')]
df_performance = df_performance.loc[:, ~df_performance.columns.str.contains('^Unnamed')]
df_shipment = df_shipment.loc[:, ~df_shipment.columns.str.contains('^Unnamed')]

print("semua data sudah clean tanpa anomali")
```

```python
print("=== Rata-rata Delay (Jam) & Total Klaim Kerusakan per Vendor ===")
#kelompokkan data sesuai 'carrier'
if 'carrier' in df_customer.columns:
  summary_carier = df.performance.groupby('carrier').agg({
      'delay_hours_avg': 'mean',
      'dmg_claims_counts': 'sum'
  }).reset_index()
  print(summary_carier.to_string(index=False))
else:
  print("Kolom 'carrier' tidak ditemukan, periksa kembali penulisan huruf besar/kecilnya!")
```

```python
# Ekspor ke format CSV tanpa menyertakan indeks baris
df_customer.to_csv('customer.xlsx - customer.csv', index=False)
df_performance.to_csv('customer.xlsx - logistics_performance.csv', index=False)
df_shipment.to_csv('customer.xlsx - shipment.csv', index=False)

print("✅ Pecahan 3 file CSV berhasil dibuat di panel kiri Google Colab!")
```
